```python
x= customer_data.iloc[:,[3,4]].values
```

```python
# calculate WCSS for different number of customers
wcss =[]

for i in range(1,11):
    kmeans=KMeans(n_clusters=i,init='k-means++',random_state =42)
    kmeans.fit(x)
    wcss.append(kmeans.inertia_)
```

```python
# plot elbow graph
plt.figure(figsize=(10,6))
plt.plot(range(1,11),wcss, marker='o',linestyle='--',color ='#e74c3c',linewidth =2,markersize =10)
plt.title('The Elbow Method',fontsize =14, fontweight='bold')
plt.xlabel('Number of clusters')
plt.ylabel('WCSS(width-clusters sum of square)')
plt.xticks(range(1,11))
plt.grid(True,alpha =0.3)
plt.show()

```

```python
# train with optimal clusters
kmeans = KMeans(n_clusters=5,init='k-means++',random_state=42)
y = kmeans.fit_predict(x)
```

```python
plt.figure(figsize=(12, 8))

colors = ['#2ecc71', '#e74c3c', '#f39c12', '#9b59b6', '#3498db']

clusters_name = [
    'standard Customers',
    'Target Customers',
    'Careless Customers',
    'Sensible Customers',
    'Careful Customers'
]

# Plot each customer cluster
for i in range(5):
    plt.scatter(
        x[y == i, 0],
        x[y == i, 1],
        s=100,
        c=colors[i],
        label=f'Cluster {i + 1}: {clusters_name[i]}',
        alpha=0.6,
        edgecolors='black'
    )

# Plot all centroids once
plt.scatter(
    kmeans.cluster_centers_[:, 0],
    kmeans.cluster_centers_[:, 1],
    s=300,
    c='black',
    marker='*',
    label='Centroids',
    edgecolors='yellow',
    linewidth=2
)

# Chart formatting
plt.title('Customer Segments: K-Means Clustering',
          fontsize=16, fontweight='bold')
plt.xlabel('Annual Income (k$)', fontsize=12)
plt.ylabel('Spending Score (1-100)', fontsize=12)

plt.legend(fontsize=10)
plt.grid(True, alpha=0.3)
plt.show()
```

```python
customer_data['Cluster'] = y

cluster_profile = customer_data.groupby('Cluster').agg(
    Customer_Count=('CustomerID', 'count'),
    Average_Age=('Age', 'mean'),
    Average_Income=('Annual Income (k$)', 'mean'),
    Average_Spending=('Spending Score (1-100)', 'mean')
).round(2)

print(cluster_profile)
```

```python
customer_data['Cluster']=y
```

```python
# create comparison visulaization
fig, axes = plt.subplots(1,2,figsize=(14,5))

#income comparison
cluster_income =[customer_data[customer_data['Cluster']==i]['Annual Income (k$)'].mean()
                 for i in range(5)]
axes[0].bar(range(5),cluster_income,color = colors)
axes[0].set_title('Average Income by Cluster', fontweight ='bold')
axes[0].set_xlabel('cluster')
axes[0].set_ylabel('Annual Income (k$)')
axes[0].set_xticks(range(5))
axes[0].set_xticklabels([f'c{i}' for i in range(5)])

# spending comparison
cluster_spending =[customer_data[customer_data['Cluster']==i]['Spending Score (1-100)'].mean()
                   for i in range(5)]
axes[1].bar(range(5),cluster_spending,color = colors)
axes[1].set_title('Average Spending Score by Cluster', fontweight ='bold')
axes[1].set_xlabel('cluster')
axes[1].set_ylabel('Spending Score')
axes[1].set_xticks(range(5))
axes[1].set_xticklabels([f'c{i}' for i in range(5)])

plt.show()









```

```python
# Example: Predict a new customer

new_customer = np.array([[70, 80]])  # Income: $70k, Spending Score: 80

predicted_customer = kmeans.predict(new_customer)[0]

print(
    f"Customer (Income: ${new_customer[0][0]}k, "
    f"Spending Score: {new_customer[0][1]}) "
    f"-- Cluster {predicted_customer}: "
    f"{clusters_name[predicted_customer]}"
)
```
